# 07 - Iceberg vs Delta

## What you'll learn
- How Iceberg and Delta differ in protocol and catalog model.
- Why field IDs and hidden partitioning matter in interviews.
- How to give a balanced answer without sounding tribal.

## Interview questions this answers
- Why would you choose Iceberg over Delta?
- When would you not choose Iceberg?

| Topic | Iceberg | Delta |
| --- | --- | --- |
| Transaction log | Metadata JSON plus Avro manifests | `_delta_log` JSON/checkpoint log |
| Schema evolution | Field IDs are first-class | Name/id mapping depends on mode and features |
| Partitioning | Hidden transforms in metadata | More directory/log oriented historically |
| Catalog | Explicit catalog abstraction | Often path/table registration oriented |
| Multi-engine story | Designed around open table spec | Strong in Databricks ecosystem, broader support improving |

If asked in interview 'why Iceberg over Delta for SLH', here's the 90-second answer:

> I would choose Iceberg when the architecture needs strong multi-engine access, explicit catalog semantics, safe schema evolution through field IDs, and hidden partitioning that does not leak physical layout into user queries. Delta can be an excellent choice in a Databricks-centered stack, but for an open lakehouse that needs Trino, Python, object storage, and future engine flexibility, Iceberg gives me a cleaner interoperability story.
