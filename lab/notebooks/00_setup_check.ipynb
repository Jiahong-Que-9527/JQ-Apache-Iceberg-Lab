```python
import boto3
import duckdb
import pandas as pd
import pyarrow as pa
import pyiceberg

from src.catalog_helper import get_catalog, get_s3_client

catalog = get_catalog()
s3 = get_s3_client()
```

The bucket check proves that the notebook container can reach MinIO through the Docker network name `minio`.

```python
import os

bucket = os.environ.get("MINIO_BUCKET", "warehouse")
buckets = [item["Name"] for item in s3.list_buckets()["Buckets"]]
print("Buckets:", buckets)
assert bucket in buckets, f"Missing bucket: {bucket}"

print("PyIceberg:", pyiceberg.__version__)
print("DuckDB:", duckdb.__version__)
print("PyArrow:", pa.__version__)
print("pandas:", pd.__version__)
print("Catalog namespaces:", list(catalog.list_namespaces()))
print("Setup OK - proceed to 01_basics")
```
