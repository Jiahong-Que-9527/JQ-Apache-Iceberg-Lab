# 03 - Schema Evolution

## What you'll learn
- Why Iceberg tracks columns by field ID.
- What metadata changes when a column is added, renamed, or dropped.
- Why old data files do not need to be rewritten for many schema changes.

## Interview questions this answers
- Why is Iceberg schema evolution safer than raw Parquet evolution?
- What is a field ID?

```python
from src.catalog_helper import current_metadata_location, get_catalog, read_json_s3

catalog = get_catalog()
table = catalog.load_table("lab.events")
metadata = read_json_s3(current_metadata_location(table))
metadata["schemas"]
```

The important thing to observe first is that every field has an `id`. Renames change names; readers still resolve fields by ID.

PyIceberg's schema update API can change across versions, so use this notebook as an inspection-first guide. Try guarded update calls only after `01_basics` is green.

```python
print("Current metadata:", current_metadata_location(table))
print("Current schema id:", metadata["current-schema-id"])
for schema in metadata["schemas"]:
    print("schema-id", schema["schema-id"], "fields", [(field["id"], field["name"]) for field in schema["fields"]])
```
